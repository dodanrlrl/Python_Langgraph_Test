## 반복
- `while`
- `for`

### `while` 구문

```python
numbers = [1, 2, 3, 4, 5]

total = 0
idx = 0

# idx  가 numbers 의 길이(len()) 보다 작다면, 아래 블럭 반복
while idx < len(numbers):
  total += numbers[idx]
  idx += 1

print('합계:', total)
```

### `for .. in` 구문

```python
numbers = [1, 2, 3, 4, 5]
total = 0

for num in numbers:
  total += num

print('합계:', total)
```

```python
#1부터 10000까지, 3의 배수를 다 더해보자
num = 1
total_1 = 0
total_2 = 0
total_3 = 0

#1
while num <= 10000: # 1e4 = 1 x 10의4제곱
  if num % 3 == 0:
    total_1 += num
  num += 1

print('합계1:', total_1)

#2
for num in range(1, 10001):
  if num % 3 == 0:
    total_2 += num

print('합계2:', total_2)

#3
for num in range(0, 10001, 3):
  total_3 += num

print('합계3:', total_3)


```
