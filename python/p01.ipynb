## 점수 평균과 합격 판별

학생 점수 리스트가 있다.

```py
scores = [80, 75, 90, 60, 100]
```
평균을 구하시오.
평균이 80 이상이면 "합격" 아니면 "불합격" 출력

````python
scores = [80, 75, 90, 60, 100]

total = sum(scores)
avg = total / len(scores)

if avg >= 80:
    print('합격')
else:
    print('불합격')  
````

## 최대값 찾기

숫자 리스트가 있다.
```
numbers = [3, 7, 2, 9, 5]
```
가장 큰 값을 찾아 출력하시오.

Tip: `max()` 없이 직접 비교해보기

````python
numbers = [3, 7, 2, 9, 5]
max = numbers[0]

for num in numbers:
    if num > max:
        max = num
print("최대값:", max)

#max(numbers) -> 내장함수로도 구할 수 있음
````

## 짝수만 저장하기
```
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
```
짝수만 새로운 리스트에 담아서 출력하시오.

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
even_numbers = []

for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)

print(even_numbers)              
````

## 이름 존재 여부 확인
```
names = ["철수", "영희", "민수", "지영"]
```
변수 target 에 저장된 이름이 리스트 안에 있으면
"존재합니다" 아니면 "없습니다" 출력

TIP: `in` 연산자 없이 풀기

````python
target = '지영'
names = ["철수", "영희", "민수", "지영"]
is_found = False
idx = 0

#1
if target in names:
    print('존재합니다')
else:
    print('없습니다')

#2
while idx < len(names):
    if names[idx] == target:
        is_found = True
    idx += 1

if is_found:
    print('존재합니다')
else:
    print('없습니다')

# in 연산자 있을때
for name in names:
    if name == target:
        is_found = True
        break

if is_found:
    print('존재합니다')
else:
    print('없습니다')


# in 연산자 없이
while idx < len(names):
    if names[idx] == target:
        print('존재합니다')
        break
    else:
        idx += 1
        if idx == len(names):
            print('없습니다')
````

## 리스트 안 딕셔너리 평균 구하기
```
students = [
    {"name": "철수", "score": 80},
    {"name": "영희", "score": 90},
    {"name": "민수", "score": 70}
]
```
전체 평균 점수 구하기
평균 이상인 학생 이름 출력

````python
students = [
    {"name": "철수", "score": 80},
    {"name": "영희", "score": 90},
    {"name": "민수", "score": 70}
]

avg = 0
avg_list = []

# 평균 구하기
for student in students:
    avg += student["score"]
avg /= len(students)

print("평균 점수:", avg)   

# 평균 이상인 학생 출력
for student in students:
    if student["score"] >= avg:
        avg_list.append(student["name"])

print(avg_list)
````

## 비밀번호 검사기

`password = "abc123"`

조건:

1. 길이가 8자 이상이고
1. 숫자가 포함되어 있으면 "사용 가능"
1. 아니면 "사용 불가"

출력하시오.

(힌트: `len()`, `in` 사용)

````python
password = "12345678"
is_valid = False
char = '1234567890'

#1
for pw in password:
    if pw in char:
        is_valid = True
        break

if len(password) >= 8 and is_valid:
    print("사용 가능") 
else:
    print("사용 불가")


#2
for char in password:
    if char.isdigit():
        is_valid = True
        break
    
if len(password) >= 8 and is_valid:
    print("사용 가능")
else:
    print("사용 불가")  
````

## 장바구니 총 금액 계산
```py
cart = [
    {"item": "노트북", "price": 900000, "qty": 1},
    {"item": "마우스", "price": 20000, "qty": 2},
    {"item": "마우스", "price": 30000, "qty": 2},
    {"item": "마우스", "price": 10000, "qty": 2},
]
```
총 결제 금액 계산
100만원 이상이면 "고액 구매자" 출력

````python
cart = [
    {"item": "노트북", "price": 900000, "qty": 1},
    {"item": "마우스", "price": 20000, "qty": 2},
    {"item": "마우스", "price": 30000, "qty": 2},
    {"item": "마우스", "price": 10000, "qty": 2},
]

total_price = 0

for price in cart:
    total_price += price["price"] * price["qty"]
    
print("총 구매 금액:", total_price)
if total_price >= 1000000:
    print("고액구매자")
````

## 등급 매기기

점수 리스트:
```py
scores = [95, 82, 67, 74, 88]
```
각 점수에 대해

- 90 이상 → 'A'
- 80 이상 → 'B'
- 70 이상 → 'C'
- 그 외 → 'D'

등급 리스트를 새로 만들어 출력하시오.

````python
scores = [95, 82, 67, 74, 88]
grades = []
for score in scores:
    if score >= 90:
        grades.append("A")
    elif score >= 80:
        grades.append("B")
    elif score >= 70:
        grades.append("C")
    else:
        grades.append("D")

print(grades)
````
