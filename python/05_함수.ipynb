````python
# name: plus_one / in: 숫자 / out: 숫자 + 1

# 함수 정의(define)

def plus_one(num):
  return num + 1

plus_one(99)
print(plus_one(99))
````

````python
# 3제곱 해주는 함수
# name: cube, in: 숫자 1개, out: 숫자의 세제곱

def cube(num):
  return num**3

print(cube(5))
````

````python
# 원기둥의 부피를 구하는 함수
# name: cylinder, in: 반지름/높이, out: 부피(pi * r^2 * h)

def cylinder(r, h):
  return Math.pi*r**2*h
  
````

````python
# 리스트 내부의 숫자들의 평균을 구해주는 함수
# name: get_avg, in: 숫자들로 만들어진 리스트, out: 평균
nums = [1,3,4,5,6]
def get_avg(numbers):
  return sum(numbers)/len(numbers)

print(get_avg(nums))
````

````python
# 비밀번호 체크해주는 함수 (숫자 포함, 8글자 이상)
# name: check_pw, in: 비밀번호 문자열, out: True/False

def check_pw(string):
    for char in string:
      if len(string) >= 8 and char.isdigit() == True:
         return True
      else:
         return False


print(check_pw("가나다라마바사"))

````

````python
# BMI 기준으로, 비만여부 확인해주는 함수 (BMI = 체중(kg) / 키(m)^2)
# 남성 27 이상일 경우 비만
# 여성 25 이상일 경우 비만

def check_fat(bmi, gender):
  if gender == 'man' and bmi >= 27:
    return True
  elif gender == 'female' and bmi >= 25:
    return True
  else:
    return False
  




````
