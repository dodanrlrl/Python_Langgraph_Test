```python
# 원 (Circle)
import math
#python에서 클래스 이름은 pascalCase를 사용(my_good_man -> 변수명(스네이크 케이스), MYGoodMan -> 클래스명(파스칼 케이스))
class Circle:
    def __init__(self,r):
        self.r = r

    def get_area(self):
        return round(math.pi*self.r**2,2)
    
    def get_perimeter(self):
        return round(2*math.pi*self.r,2)      

c1 = Circle(3)
c2 = Circle(5)

c1.get_area()
```
